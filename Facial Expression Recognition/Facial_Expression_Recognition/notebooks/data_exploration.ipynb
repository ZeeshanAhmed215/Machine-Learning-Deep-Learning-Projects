# FER-2013 data exploration
Run this notebook **from the `notebooks/` folder** after placing the dataset in `dataset/` (see `dataset/README.md`). No outputs are stored, so everything you see comes from your own data.

```python
import sys
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT))

import numpy as np
import matplotlib.pyplot as plt
from src import config
from src.data_loader import load_raw_dataset, inspect_dataset, class_distribution, make_splits
```

```python
raw = load_raw_dataset(config.DATASET_DIR)
print('Layout:', raw.layout, '| images:', len(raw.images))
print('Corrupt files:', len(raw.report['corrupt_files']), '| invalid CSV rows:', raw.report['invalid_rows'])
```

```python
summary = inspect_dataset(raw)
print('Exact duplicate images:', summary['exact_duplicate_images'])
table = class_distribution(raw)
table
```

```python
import seaborn as sns
ax = table['total'].rename(index=str.capitalize).plot.bar(figsize=(8, 4), color=sns.color_palette('viridis', 7))
ax.set_ylabel('Images'); ax.set_title('Class distribution'); plt.show()
```

```python
fig, axes = plt.subplots(7, 6, figsize=(9, 11))
for row, name in enumerate(config.DISPLAY_NAMES):
    idx = np.flatnonzero(raw.labels == row)[:6]
    for col in range(6):
        axes[row, col].axis('off')
        if col < len(idx):
            axes[row, col].imshow(raw.images[idx[col]], cmap='gray')
        if col == 0:
            axes[row, col].set_title(name, loc='left', fontsize=10)
plt.tight_layout(); plt.show()
```
